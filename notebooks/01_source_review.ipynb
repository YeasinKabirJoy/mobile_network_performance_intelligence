# Phase 1 - Source review

This notebook presents reproducible evidence from the official Q4 2024 Ookla mobile Parquet source. It verifies schema, quality, and Bangladesh availability without creating the Phase 2 extracted dataset.

```python
from pathlib import Path
import json
import subprocess
import sys

project_root = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
output = project_root / 'reports/generated/source_inspection.json'
subprocess.run([sys.executable, '-m', 'src.inspect_source'], cwd=project_root, check=True, stdout=subprocess.DEVNULL)
inspection = json.loads(output.read_text(encoding='utf-8'))
```
